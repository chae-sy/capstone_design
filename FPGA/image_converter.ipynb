```python
import numpy as np
from PIL import Image

def image_to_coe_4bit(image_path: str, coe_path: str):
    # 1) 이미지 읽기 및 크기 검증
    img = Image.open(image_path).convert('RGB')
    w, h = img.size
    assert (w, h) == (100, 100), f"Expected 100×100 image, but got {w}×{h}"
    
    # 2) NumPy 배열로 변환 (shape = [100,100,3])
    data = np.array(img, dtype=np.uint8)
    
    # 3) .coe 파일 쓰기
    total_pixels = w * h
    count = 0
    
    with open(coe_path, 'w') as f:
        f.write('memory_initialization_radix=16;\n')
        f.write('memory_initialization_vector=\n')
        
        for y in range(h):
            for x in range(w):
                r, g, b = data[y, x]
                # 상위 4비트만 취하기
                r4 = r >> 4
                g4 = g >> 4
                b4 = b >> 4
                # 12비트 값으로 결합 (R[11:8], G[7:4], B[3:0])
                val12 = (r4 << 8) | (g4 << 4) | b4
                hex_str = f"{val12:03X}"  # 3자리 HEX
                
                count += 1
                # 마지막 픽셀은 세미콜론, 아니면 콤마
                sep = ';' if count == total_pixels else ','
                f.write(hex_str + sep)
            
            f.write('\n')
    
    print(f"Generated 4-bit COE file: {coe_path}")



```

```python
image_to_coe_4bit('1_ori.png', 'ori_1.coe')
image_to_coe_4bit('2_ori.png', 'ori_2.coe')
image_to_coe_4bit('3_ori.png', 'ori_3.coe')
image_to_coe_4bit('1_sr.png', 'sr_1.coe')
image_to_coe_4bit('2_sr.png', 'sr_2.coe')
image_to_coe_4bit('3_sr.png', 'sr_3.coe')
```
